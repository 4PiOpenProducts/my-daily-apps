```python
import os
import subprocess
import glob

# 1. Google ドライブのマウント
from google.colab import drive
if not os.path.exists('/content/drive'):
    print("Googleドライブをマウント中...")
    drive.mount('/content/drive')

# 2. 目的のフォルダ名（今回のプロジェクトフォルダ）
target_folder = "day2-templates"  # ※必要に応じて "day2-template" などに合わせる

# 3. /content 配下や Googleドライブ内から該当するフォルダを全検索して自動移動する
print(f"'{target_folder}' フォルダを検索中...")
found_paths = glob.glob(f"/content/**/{target_folder}", recursive=True)

if found_paths:
    # 見つかった最初のパスに移動
    project_dir = found_paths[0]
    os.chdir(project_dir)
    print(f"✨ フォルダを発見し、移動しました: {os.getcwd()}")
else:
    # 見つからない場合は現在の場所を維持
    print(f"⚠️ フォルダが見つかりませんでした。現在の場所で続行します: {os.getcwd()}")

# 4. uv が未インストールの場合のみインストール
if subprocess.run(["which", "uv"], capture_output=True).returncode != 0:
    print("uv をインストール中...")
    subprocess.run(["pip", "install", "-q", "uv"], check=True)

# 5. uv を使ってプロジェクトの依存関係をシステム環境に高速同期
print("uv を使って環境を構築中...")
result = subprocess.run(["uv", "pip", "install", "--system", "-e", "."], capture_output=True, text=True)

if result.returncode == 0:
    print("\n✨ セットアップ完了！環境構築に成功しました。")
else:
    print("\n❌ エラーが発生しました:")
    print(result.stderr)
```
